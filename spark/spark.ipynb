```python
import os
import subprocess

print("JAVA_HOME =", os.environ.get("JAVA_HOME"))

result = subprocess.run(
    ["java", "-version"],
    capture_output=True,
    text=True
)

print("JAVA VERSION:")
print(result.stderr)

print("JAVA PATH:")
result = subprocess.run(
    ["where", "java"],
    capture_output=True,
    text=True
)
print(result.stdout)
```
