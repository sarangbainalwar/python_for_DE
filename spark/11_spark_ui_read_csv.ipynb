```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("MySpark").getOrCreate()
```

```python
spark
```

```python
df=spark.read.format("csv").load("emp.csv")
```

```python
df.printSchema()
```

```python
df=spark.read.format("csv").option("header",True).load("emp.csv")
```

```python
df.printSchema()
```

```python
df=spark.read.format("csv").option("header",True).option("inferSchema",True).load("emp.csv")
```

```python
df.printSchema()
```

```python
df.show()
```

```python
_schema = "employee_id int,department_id int,name string,age int,gender string,salary double,hire_date date"

df_schema=spark.read.format("csv").option("header",True).schema(_schema).load("emp.csv")
```

```python
df_schema.show()
```

```python
_schema = "employee_id int,department_id int,name string,age int,gender string,salary double,hire_date date,columnNameOfCorruptRecord string"

df_p=spark.read.format("csv").schema(_schema).option("columnNameOfCorruptRecord", "bad_record").option("header",True).load("emp_new.csv")
```

```python
df_p.show()
```
