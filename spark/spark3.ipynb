```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Practice').getOrCreate()
```

```python
df_pyspark=spark.read.csv('test2.csv',header=True,inferSchema=True)
```

```python
df_pyspark.show()
```

```python
### drop the columns
df_pyspark.drop('Name').show()
```

```python
df_pyspark.show()
```

```python
df_pyspark.na.drop().show()
```

```python
### how=all: drops the rows where all the column values are null
### how=any: drops the rows where even anyone of the column values are null
df_pyspark.na.drop(how="all").show()
df_pyspark.na.drop(how="any").show()
```

```python
### threshold tells the minimum numbr of non null values to be present in the row to not be deleted as null
df_pyspark.na.drop(how="any",thresh=2).show()
```

```python
# subset: specifies the column name from which null should be deleted
df_pyspark.na.drop(how="any",subset=['experience']).show()
```

```python
# fill the missing values
df_pyspark.na.fill('missing values').show()
df_pyspark.na.fill(999,['experience','age']).show()
```

```python
df_pyspark.show()
```

```python
# replace nullls with mean of experience
from pyspark.ml.feature import Imputer
imputer = Imputer(
    inputCols=['age','Experience','Salary'],
    outputCols=["{}_imputed".format(c) for c in ['age','Experience','Salary']]
).setStrategy("mode")
```

```python
imputer.fit(df_pyspark).transform(df_pyspark).show()
```
